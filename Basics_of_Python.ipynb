```python
# Example of append()
my_list_append = [1, 2, 3]
print(f"Initial list for append: {my_list_append}")

my_list_append.append(4) # Appends a single element
print(f"After my_list_append.append(4): {my_list_append}")

my_list_append.append([5, 6]) # Appends the list [5, 6] as a single element
print(f"After my_list_append.append([5, 6]): {my_list_append}")

my_list_append.append('abc') # Appends the string 'abc' as a single element
print(f"After my_list_append.append('abc'): {my_list_append}")
```

#### Methods applicable to various Python data types:

*   **Strings** (`str`)
    *   `.upper()`: Returns a copy of the string with all the cased characters converted to uppercase.
    *   `.lower()`: Returns a copy of the string with all the cased characters converted to lowercase.
    *   `.strip()`: Returns a copy of the string with leading and trailing whitespace removed.
    *   `.split(separator)`: Returns a list of substrings separated by the given delimiter.
    *   `.replace(old, new)`: Returns a copy of the string with all occurrences of `old` replaced by `new`.

*   **Lists** (`list`)
    *   `.append(element)`: Adds a single element to the end of the list.
    *   `.extend(iterable)`: Appends elements from an iterable to the end of the list.
    *   `.insert(index, element)`: Inserts an element at a specified position.
    *   `.remove(element)`: Removes the first occurrence of a specified element.
    *   `.pop(index)`: Removes and returns the element at the given index (or the last element if no index is specified).
    *   `.sort()`: Sorts the list in ascending order in-place.
    *   `.reverse()`: Reverses the order of elements in the list in-place.

*   **Tuples** (`tuple`)
    *   `.count(value)`: Returns the number of times a specified value occurs in the tuple.
    *   `.index(value)`: Searches the tuple for a specified value and returns the position of where it was found.

*   **Sets** (`set`)
    *   `.add(element)`: Adds a single element to the set.
    *   `.remove(element)`: Removes a specified element from the set; raises a `KeyError` if the element is not found.
    *   `.discard(element)`: Removes a specified element from the set if it is present.
    *   `.pop()`: Removes and returns an arbitrary element from the set.
    *   `.union(other_set)` or `|`: Returns a new set containing all elements from both sets.
    *   `.intersection(other_set)` or `&`: Returns a new set containing only the elements common to both sets.

*   **Dictionaries** (`dict`)
    *   `.keys()`: Returns a view object that displays a list of all the keys in the dictionary.
    *   `.values()`: Returns a view object that displays a list of all the values in the dictionary.
    *   `.items()`: Returns a view object that displays a list of a dictionary's key-value tuple pairs.
    *   `.get(key, default)`: Returns the value for the specified key if it's in the dictionary, otherwise returns `default`.
    *   `.pop(key, default)`: Removes the specified key and returns its value. If the key is not found, `default` is returned if provided, otherwise a `KeyError` is raised.

```python
def check_even_list(num_list):

    even_numbers = []

    # Go through each number
    for number in num_list:
        # Once we get a "hit" on an even number, we append the even number
        if number % 2 == 0:
            even_numbers.append(number)
        # Don't do anything if its not even
        else:
            pass
    # Notice the indentation! This ensures we run through the entire for loop
    return even_numbers
```

#### Question 2 :
##### Given a list having Names, work_hours, and gender, Write a function to print name of the female worker that worked the most hours. Also how much do should she be paid if the pay is $ 20 per hour.

##### work_hours = [('Abby',100 , 'F'),('Billy',400, 'M'),('Cassie',800,'F'), ('Maggi',600,'F'),('Alex',500,'M'),('Raj',225,'M'),('Penny',920,'F'),('Ben',300,'M')]

##### Answer : the female worker that worked the most hours is Penny and she should be paid 18400

```python
## Q1:
def reverse_concate_return(a:str, b:str) -> str:
  reverse_first = a[::-1]
  return reverse_first + b

# Example usage with the modified function
a = "olleH"
b = " World"
result = reverse_concate_return(a, b)
print(f"The result from the returning function is: {result}")

# You can then use 'result' for other operations
# print(result.upper())
```

```python
## Q2:
work_hours = [('Abby',100 , 'F'),('Billy',400, 'M'),('Cassie',800,'F'), ('Maggi',600,'F'),('Alex',500,'M'),('Raj',225,'M'),('Penny',920,'F'),('Ben',300,'M')]
def highest_paid_female_worker(work_hours):
    max_hour = 0
    female_worker = None

    for name, hours, gender in work_hours:
        if gender == 'F' and hours > max_hour:
            max_hour = hours
            female_worker = name
            print(female_worker)

    if female_worker:
        pay = max_hour * 20
        return f"The female worker that worked the most hours is {female_worker} and she should be paid {pay}"

    else:
        return "No female workers found in the list."

result = highest_paid_female_worker(work_hours)
print(result)
```

```python
a_range = np.arange(10,19).reshape(3,3) ## create a 3x3 array with values in range 10-19
a_range
```

```python
## np.eye(size) give the identity  matrix of size
iden = np.eye(3) #idnetity matrix of given size
iden
```

```python
# Create a 4x4 identity matrix
import numpy as np
identity_matrix_4x4 = np.eye(4)
print("4x4 Identity Matrix:")
print(identity_matrix_4x4)

# Create a 2x2 identity matrix
identity_matrix_2x2 = np.eye(2)
print("\n2x2 Identity Matrix:")
print(identity_matrix_2x2)
```

```python
## adding two matrices
summed = a_range + iden
summed
```

```python
### arrays support normal matrix multiplication that you are used to, point-wise multiplication
### and dot product as well.

mul = a_range@iden ## normal multiplication
mul
# a_range.dot(iden)   ## Dot multiplication
```

```python
## point wise multiplication
p_mul = a_range * iden
p_mul
```

```python
## Transpose of a matrix.

mtx_t = mul.T
mtx_t
```

```python
### Here we are changing the values of last row of the transposed matrix.
### basically point wise multiplying the values of last row with 1,2 and 3

mtx_t[2] = mtx_t[2]*[1,2,3] ## indexing, point wise multiplication and mutation of values
mtx_t
```

```python
## Just like the greater than 1 (a>1) example we saw earlier.
## here we are checking if the elements are divisible by 2 (%), and if they are, then replace by 0.

mtx_t[(mtx_t % 2 == 0)] = 0 ## convert even elements of the matrix to zero.
mtx_t
```

```python
## A
## using hStack
# 🔹 np.hstack (Horizontal Stack)
# Stacks arrays column by column (side by side).

# Think of it as adding more columns
col =np.arange(1,6).reshape(5,1)
matrix = np.hstack([col]*5)
print(matrix)
```

```python
## A
## Using Vstack

import numpy as np
matrix = np.vstack([np.array([i]*5) for i in range(1,6)])
print(matrix)

```
